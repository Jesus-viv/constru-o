# Valores nulos 

```python
print(df.isnull().sum())

# Numéricas: mediana. Categóricas: moda (ou "Não informado")
for col in df.columns:
    if df[col].isnull().any():
        if df[col].dtype == "object":
            df[col] = df[col].fillna(df[col].mode()[0])
        else:
            df[col] = df[col].fillna(df[col].median())
```
