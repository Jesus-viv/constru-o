```python

import pandas as pd
dados = {
    "Nome": ["Ana", "Bruno", "Daniel", "Luciano"],
    "Idade": [20, 25, 30, 35],
    "Altura": [1.65, 1.80, 1.75, 1.70],
    "Curso": ["Engenharia", "Medicina", "Direito", "Arquitetura"],
    "Escolaridade": ["Superior", "Superior", "Superior", "Superior"],
    "Aprovado": [True, True, False, True],
    "Data_Matricula": pd.to_datetime(["2020-01-15", "2019-08-20", "2021-03-10", "2018-11-05"])
}

df = pd.DataFrame(dados)
```

```python
# Tipos de Variáveis
## Numéricas: int, float

    - Discretas: assumemvalores inteiros (ex: Idade)
    - Contínuas: podem assumir qualquer valor dentro de um intervalo (ex: Altura, peso, temperatura)


## Categóricas: str, bool
```

```python

print("Idade (variável numérica discreta):")
print(df["Idade"])

print("Altura (variável numérica contínua):")
print(df["Altura"])

```

```python
# Variaveis Categoricas 

- Nominais: não possuem ordem ou hierarquia (ex: Curso, Escolaridade)
- Ordinais: possuem ordem ou hierarquia (ex: Escolaridade, Nível de Satisfação)
```

```python

print("Curso (variável categórica nominal):")
print(df["Curso"])

print("Escolaridade (variável categórica ordinal):")
print(df["Escolaridade"])
```

```python
#Variavel Binarias

São variáveis que podem assumir apenas dois valores, geralmente representando uma característica booleana (ex: Sim/Não, Verdadeiro/Falso).

print("Aprovado (variável binária):")
print(df["Aprovado"])
```

```python
# Variaveis Temporais

Representam datas e horários, permitindo análises relacionadas ao tempo (ex: Data de Matrícula, Data de Nascimento).

print("Data de Matrícula (variável temporal):")
print(df["Data_Matricula"])
```
