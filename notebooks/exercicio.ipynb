# 1. Colunas numéricas

idade, nota_final, frequencia_pct, satisfacao, horas_estudo_semana

Preencher com a mediana (mais robusta contra outliers):

```python
df["idade"] = df["idade"].fillna(df["idade"].median())

df["nota_final"] = df["nota_final"].fillna(df["nota_final"].median())

df["frequencia_pct"] = df["frequencia_pct"].fillna(
    df["frequencia_pct"].median()
)
```

#Colunas categóricas

cidade, curso, modalidade, dispositivo_principal

```python
df["cidade"] = df["cidade"].fillna(
    df["cidade"].mode()[0]
)

df["curso"] = df["curso"].fillna(
    df["curso"].mode()[0]
)
```

#riar gráficos com Matplotlib: histograma de notas, barras por curso, dispersão entre frequência e nota, e evolução de matrículas por mês.

```python
# Histograma das notas finais
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['nota_final'], bins=10, edgecolor='black')
plt.title('Distribuição das Notas Finais')
plt.xlabel('Nota Final')
plt.ylabel('Quantidade de Alunos')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['frequencia_pct'],
    df['nota_final'],
    alpha=0.7
)

plt.title('Frequência x Nota Final')
plt.xlabel('Frequência (%)')
plt.ylabel('Nota Final')
plt.grid(True, alpha=0.3)
plt.show()
# Gráfico de barras por curso
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['frequencia_pct'],
    df['nota_final'],
    alpha=0.7
)

plt.title('Frequência x Nota Final')
plt.xlabel('Frequência (%)')
plt.ylabel('Nota Final')
plt.grid(True, alpha=0.3)
plt.show()
# Gráfico de dispersão entre frequência e nota
```

```python
df['data_matricula'] = pd.to_datetime(
    df['data_matricula'],
    errors='coerce'
)
#Evolução das matrículas por mês
```

```python
matriculas_mes = (
    df.groupby(df['data_matricula'].dt.to_period('M'))
      .size()
)

matriculas_mes.index = matriculas_mes.index.astype(str)

plt.figure(figsize=(10,5))

plt.plot(
    matriculas_mes.index,
    matriculas_mes.values,
    marker='o'
)

plt.title('Evolução das Matrículas por Mês')
plt.xlabel('Mês')
plt.ylabel('Quantidade de Matrículas')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.show()
```
